# Rule-Based Chatbot

A transparent, dependency-free FAQ chatbot built on regex intents — no ML, no API keys, fully explainable.


```python

import re, random

INTENTS = [
    ("greeting", [r"\b(hi|hello|hey|salam|aoa)\b"], ["Hello! How can I help you today?", "Hi there! Ask me about services, prices or timings."]),
    ("services", [r"\b(services?|offers?|do you do|provide)\b"], ["We offer AI/ML projects, data analysis, dashboards and automation. Tell me what you need built."]),
    ("pricing", [r"\b(prices?|pricing|costs?|charges?|rates?)\b"], ["Pricing depends on scope. Small ML models start from a fixed mini-package; share your requirements for an exact quote."]),
    ("timings", [r"\b(timings?|hours|open|when)\b"], ["We reply within 24 hours, Monday to Saturday."]),
    ("contact", [r"\b(contact|email|phone|reach)\b"], ["You can reach us by email at ikramrehman300@gmail.com."]),
    ("projects", [r"\b(projects?|portfolio|github|work)\b"], ["You can explore 15 working AI projects in the portfolio, and try them live in the AI Projects Lab app."]),
    ("thanks", [r"\b(thank|shukria)\b"], ["You are welcome! Anything else I can help with?"]),
    ("bye", [r"\b(bye|goodbye|allah hafiz)\b"], ["Goodbye! Come back any time."]),
]
FALLBACKS = ["I am a simple rule-based bot, so I did not catch that. Try asking about services, pricing, timings or projects.",
             "Hmm, that is outside my rules. Ask me about services, prices or contact details."]

class ChatBot:
    def __init__(self, intents=INTENTS):
        self.intents = intents
    def reply(self, text):
        t = text.lower()
        for name, patterns, answers in self.intents:
            if any(re.search(p, t) for p in patterns):
                return name, random.choice(answers)
        return "fallback", random.choice(FALLBACKS)

if __name__ == "__main__":
    bot = ChatBot()
    script = ["Hello!", "What services do you offer?", "What are your prices?",
              "Show me your projects", "How can I contact you?", "What are your timings?",
              "Thank you so much", "Tell me a joke about quantum physics", "Bye!"]
    hits = 0
    for line in script:
        intent, ans = bot.reply(line)
        expected_fallback = "joke" in line
        ok = (intent == "fallback") == expected_fallback
        hits += ok
        print(f"You: {line}\nBot [{intent}]: {ans}\n")
    print(f"Intent check: {hits}/{len(script)} behaved as expected")
```

```python

bot = ChatBot()
script = ["Hello!", "What services do you offer?", "What are your prices?",
          "Show me your projects", "How can I contact you?", "What are your timings?",
          "Thank you so much", "Tell me a joke about quantum physics", "Bye!"]
for line in script:
    intent, ans = bot.reply(line)
    print(f"You: {line}\nBot [{intent}]: {ans}\n")
```

## How it works

Each intent has regex patterns and canned replies. The bot scans intents in order and returns the first match, otherwise a fallback. Simple, transparent and dependency-free — the classic pre-LLM chatbot design, and still the right tool for FAQs.
