# Fashion-MNIST Clothing Classification

A neural network that recognises 10 types of clothing from 28x28 images (Fashion-MNIST, 70,000 images).


```python

import gzip, io, urllib.request
import numpy as np
def load_idx_images(url):
    req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
    raw = gzip.decompress(urllib.request.urlopen(req).read())
    n = int.from_bytes(raw[4:8], "big"); rows = int.from_bytes(raw[8:12], "big"); cols = int.from_bytes(raw[12:16], "big")
    return np.frombuffer(raw[16:], dtype=np.uint8).reshape(n, rows * cols)
def load_idx_labels(url):
    req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
    raw = gzip.decompress(urllib.request.urlopen(req).read())
    return np.frombuffer(raw[8:], dtype=np.uint8)
BASE_URL = "https://storage.googleapis.com/tensorflow/tf-keras-datasets/"
x_train = load_idx_images(BASE_URL + "train-images-idx3-ubyte.gz")
y_train = load_idx_labels(BASE_URL + "train-labels-idx1-ubyte.gz")
x_test = load_idx_images(BASE_URL + "t10k-images-idx3-ubyte.gz")
y_test = load_idx_labels(BASE_URL + "t10k-labels-idx1-ubyte.gz")
labels = ["T-shirt/top","Trouser","Pullover","Dress","Coat","Sandal","Shirt","Sneaker","Bag","Ankle boot"]
print(x_train.shape, x_test.shape)
```

```python

from sklearn.neural_network import MLPClassifier
X_train = x_train[:20000] / 255.0
y_tr = y_train[:20000]
X_test = x_test / 255.0
mlp = MLPClassifier(hidden_layer_sizes=(128,), max_iter=25, random_state=42)
mlp.fit(X_train, y_tr)
print("Test accuracy:", round(mlp.score(X_test, y_test), 4))
```

```python

preds = mlp.predict(X_test[:8])
for i in range(8):
    print(f"true={labels[y_test[i]]:12s} predicted={labels[preds[i]]}")
```
