# News Topic Classification (AG News)

Classifies news articles into World / Sports / Business / Sci-Tech (AG News, 120,000 training articles) with TF-IDF + logistic regression.


```python

import pandas as pd
tr = pd.read_parquet("https://huggingface.co/datasets/fancyzhx/ag_news/resolve/main/data/train-00000-of-00001.parquet")
te = pd.read_parquet("https://huggingface.co/datasets/fancyzhx/ag_news/resolve/main/data/test-00000-of-00001.parquet")
NAMES = ["World", "Sports", "Business", "Sci/Tech"]
print(tr.shape, te.shape); tr.head()
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
vec = TfidfVectorizer(max_features=30000, ngram_range=(1, 2), sublinear_tf=True)
X_train = vec.fit_transform(tr["text"]); X_test = vec.transform(te["text"])
m = LogisticRegression(max_iter=1000).fit(X_train, tr["label"])
print("Test accuracy:", round(float(m.score(X_test, te["label"])), 4))
```

```python

samples = ["The stock market rose after the central bank held rates steady",
           "The striker scored twice as the home side won the derby",
           "Scientists unveil a faster chip built on a new architecture",
           "The summit ended with a joint statement on trade and security"]
for s in samples:
    print(f"{NAMES[int(m.predict(vec.transform([s]))[0])]:9s} <- {s}")
```
