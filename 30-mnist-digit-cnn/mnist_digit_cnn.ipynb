# Handwritten Digit Recognition with a CNN

**Dataset:** MNIST — 60,000 training and 10,000 test images of handwritten digits (28×28 grayscale). Project 05 solved this with a plain MLP (96.2%); this project asks what a **convolutional** network adds: weight sharing and spatial structure.

**Result: the CNN scores ~99%** — the gap between 96% and 99% is the difference between seeing pixels and seeing shapes.

```python
import tensorflow as tf
(xtr, ytr), (xte, yte) = tf.keras.datasets.mnist.load_data()
xtr = xtr.reshape(-1, 28, 28, 1) / 255.0
xte = xte.reshape(-1, 28, 28, 1) / 255.0
print(xtr.shape, xte.shape)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, 3, activation='relu', input_shape=(28, 28, 1)),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation='relu'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
history = model.fit(xtr, ytr, epochs=5, batch_size=128, validation_split=0.05)
loss, acc = model.evaluate(xte, yte, verbose=0)
print('Test accuracy:', round(float(acc), 4))
```

## What I learned
- Two small conv blocks beat a much larger MLP: convolutions reuse the same feature detectors everywhere in the image, so a stroke learned in one corner works in all corners.
- Pooling buys translation tolerance — a digit shifted by two pixels is still the same digit.
- The remaining errors are genuinely ambiguous handwriting (4/9, 3/5/8 shapes).
